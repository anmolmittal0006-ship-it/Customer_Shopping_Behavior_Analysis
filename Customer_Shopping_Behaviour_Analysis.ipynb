```python

import pandas as pd

df = pd.read_csv('/Users/Anmol/Data_Analytics_Projects /customer_shopping_behavior.csv')
```

```python
# Checking For Null Values 

df.isnull().sum()
```

```python
# Replacing Null Values With Median of same category.

df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
# Changing all column names to snake_case

df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_') 
df = df.rename(columns = {'purchase_amount_(usd)': 'purchase_amount'})
```

```python
df.columns
```

```python
# Creating Column For Age Group 
labels = ['Yonung Adult','Adult','Middle-aged','Senior']

df['age_group'] = pd.qcut(df['age'],q=4,labels=labels)

```

```python
df[['age','age_group']].head(10)
```

```python
df['frequency_of_purchases'].unique()
```

```python
# Creating column purchase_frequency_days

frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```
